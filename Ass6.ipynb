```python
import pandas as pd
from tensorflow.keras.layers import Dense,Input,Embedding,Flatten,Concatenate
from tensorflow.keras.models import Model
```

```python
movies=pd.read_csv("movies.csv")
ratings=pd.read_csv("ratings.csv")
```

```python
movies.head()
```

```python
rattings=ratings.head(2000).copy()
```

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
```

```python
rattings['user']=le.fit_transform(rattings['userId'])
rattings['movie']=le.fit_transform(rattings['movieId'])
```

```python
user_nums=rattings['user'].nunique()
movie_nums=rattings['movie'].nunique()
```

```python
user_nums
```

```python
movie_nums
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(
    rattings[['user','movie']],
    rattings['rating'],
    test_size=0.2
)
```

```python
X_train.shape
```

```python
y_train.shape
```

```python
user_input=Input(shape=(1,))
movie_input=Input(shape=(1,))

user_embedding=Embedding(user_nums,8)(user_input)
movie_embedding=Embedding(movie_nums,8)(movie_input)

user_vec=Flatten()(user_embedding)
movie_vec=Flatten()(movie_embedding)

combined=Concatenate()([user_vec,movie_vec])

hidden=Dense(32,activation='relu')(combined)
output=Dense(1)(hidden)

model=Model([user_input,movie_input],output)

model.compile(optimizer='adam',loss='mse',metrics=['mae'])
```

```python
model.fit([X_train['user'],X_train['movie']],y_train,epochs=20,batch_size=32)
```

```python
model.summary()
```

```python
pred=model.predict([X_test['user'],X_test['movie']])
```

```python
from sklearn.metrics import mean_absolute_error,root_mean_squared_error
```

```python
print("mae:",mean_absolute_error(y_test,pred))
print("RMSE",root_mean_squared_error(y_test,pred))
```
