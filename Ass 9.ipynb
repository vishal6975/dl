```python
import pandas as pd
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input,Dense,Dropout,Flatten
from tensorflow.keras.utils import to_categorical
```

```python
X_train=np.loadtxt(r"D:\dl ass prac\MNIST_Dataser-20261023T133331Z-1-001 (1)\MNIST_Dataser\UCI HAR Dataset\train\X_train.txt")
X_test=np.loadtxt(r"D:\dl ass prac\MNIST_Dataser-20261023T133331Z-1-001 (1)\MNIST_Dataser\UCI HAR Dataset\test\X_test.txt")
y_train=np.loadtxt(r"D:\dl ass prac\MNIST_Dataser-20261023T133331Z-1-001 (1)\MNIST_Dataser\UCI HAR Dataset\train\y_train.txt")
y_test=np.loadtxt(r"D:\dl ass prac\MNIST_Dataser-20261023T133331Z-1-001 (1)\MNIST_Dataser\UCI HAR Dataset\test\y_test.txt")
```

```python
activities=[
    "WALKING",
    "WALKING_UPSTAIRS",
    "WALKING_DOWNSTAIRS",
    "SITTING",
    "STANDING",
    "LAYING"
]
```

```python
y_train=y_train.astype (int) -1
y_test=y_test.astype (int) -1 
y_train=to_categorical(y_train,6)
y_test=to_categorical(y_test,6)
```

```python
dnn=Sequential([
    Input(shape=(561,)),
    Dense(256,activation='relu'),
    Dropout(0.3),
    Dense(128,activation='relu'),
    Dropout(0.3),
    Dense(56,activation='relu'),
    Dense(6,activation='softmax')
])
```

```python
dnn.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
dnn.fit(X_train,y_train,epochs=5,batch_size=32)
```

```python
loss,accuracy=dnn.evaluate(X_test,y_test)
```

```python
i=0
pred=dnn.predict(X_test[i:i+1])
print("actual:",activities[np.argmax(y_test[i])])
print("predicted:",activities[np.argmax(pred)])
```
